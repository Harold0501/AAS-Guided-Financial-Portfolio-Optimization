```python
import pandas as pd

AM = pd.read_csv("Am/merged_Am_CR.csv")
CS = pd.read_csv("merged_CS_CR.csv")
AM["instance"] = (AM["instance"] + 10000).astype(int)
ALL = pd.concat([CS, AM])
ALL.to_csv("All_CR.csv",index=False)
```
