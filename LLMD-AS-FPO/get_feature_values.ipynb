```python
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
# EPS

ticker = yf.Ticker("AAPL")
income_stmt = ticker.income_stmt
basic_eps = income_stmt.loc["Basic EPS"]
print("basic eps", basic_eps)
```

```python
# P/E

info = ticker.info

# 估值比率
pe = info.get("trailingPE", "N/A")          # P/E
pb = info.get("priceToBook", "N/A")         # P/B
ps = info.get("priceToSalesTrailing12Months", "N/A")  # P/S
p_cf = info.get("priceToCashflowTrailing12Months", "N/A")  # P/CF
ev_ebitda = info.get("enterpriseToEbitda", "N/A")  # EV/EBITDA

# 盈利能力
gross_margin = info.get("grossMargins", "N/A")  # 毛利率
net_margin = info.get("profitMargins", "N/A")   # 净利率
roa = info.get("returnOnAssets", "N/A")        # ROA
roe = info.get("returnOnEquity", "N/A")        # ROE

# 现金流
free_cash_flow = info.get("freeCashflow", "N/A")  # FCF
shares = info.get("sharesOutstanding", 1)      # 流通股数
fcf_per_share = free_cash_flow / shares if free_cash_flow != "N/A" else "N/A"  # FCF per Share

# 每股销售额 (SPS)
sps = info.get("revenuePerShare", "N/A")

# 打印结果
print(f"市盈率 (P/E): {pe}")
print(f"市净率 (P/B): {pb}")
print(f"市销率 (P/S): {ps}")
print(f"市现率 (P/CF): {p_cf}")
print(f"EV/EBITDA: {ev_ebitda}")
print(f"毛利率 (Gross Margin): {gross_margin}")
print(f"净利率 (Net Margin): {net_margin}")
print(f"资产回报率 (ROA): {roa}")
print(f"股东权益回报率 (ROE): {roe}")
print(f"自由现金流 (FCF): {free_cash_flow}")
print(f"每股自由现金流 (FCF per Share): {fcf_per_share}")
print(f"每股销售额 (SPS): {sps}")
```
